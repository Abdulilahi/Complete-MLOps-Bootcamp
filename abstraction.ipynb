```python
from abc import ABC,abstractmethod
class Vehicle(ABC):
    def drive(self):
        print("The vehicle is used for driving")
    @abstractmethod
    def start_engine(self):
        pass
class Car(Vehicle):
    def start_engine(self):
        super().drive()
        print("This car engine has been started")
    def __init__(self):
        pass
def operate_vehicle(Vehicle):
    Vehicle.start_engine()
    Vehicle.drive()

car1=Car()
operate_vehicle(car1)
```
