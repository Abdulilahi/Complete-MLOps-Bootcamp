```python
class person:
    def __init__(self,name,age,gender):
        self._name=name
        self._age=age
        self.gender=gender
class Employee(person):
    def __init__(self,name,age,gender):
        super().__init__(name,age,gender)
    def print_components(self):
        print(f"The name of the employee is {self._name}, "
              f"age is {self._age}, and is a {self.gender}")
emp1=Employee("abdul",21,"male")
emp1.print_components()
    
```
