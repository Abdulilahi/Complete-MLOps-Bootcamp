```python
import logging
logging.basicConfig(
    filename='app.log',
    filemode='w',
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%M-%D %H-%M-%S'
)
logging.debug("This is debug message")
logging.info("This is Info Message")
logging.warning("This is warning message")
logging.error("This is error message")
logging.critical("this is critical messgae")
```

```python
from logger import logging
def add(a,b):
    logging.debug("The Addition operation is taking place")
    return a+b
def sub(a,b):
    logging.debug("The subtraction operation is taking place")
    return a-b
def mul(a,b):
    logging.debug("The Multiplication operation is taking place")
    return a*b
def div(a,b):
    logging.debug("The Division operation is taking place")
    return a/b
```
