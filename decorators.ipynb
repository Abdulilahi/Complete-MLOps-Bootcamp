### Closure

```python
def main_welcome():
    msg='welcome'
    def sub_welcome_method():
        print("welcome to the advanced python course")
        print(msg)
        print("Hey!! Please learn these concepts from these")
    return sub_welcome_method()
```

```python
# Decorators


def main_welcome(func):

    def sub_welcome_method():
        print("welcome to the advanced python course")
        func()
        print("Hey!! Please learn these concepts from these")
    return sub_welcome_method()
```

```python
def course_introduction():
    print("This is the advanced python course")
```

```python
main_welcome(course_introduction)
```

```python
@main_welcome
def course_introduction():
    print("This is the advanced python course")
```

```python
def my_decorator(func):
    def wrapper():
        print("Something that is happening before function is called.")
        func()
        print("Something that is happening after function is called.")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("HELLOOOOOOOOO")
```

```python
say_hello()
```

```python
def repeat(n):
    def decorator(func):
        def wrapper(*args,**kwargs):
            for _ in range(n):
                func(*args,**kwargs)
        return wrapper
    return decorator
```

```python
@repeat(3)
def say_hello():
    print("HELLLLOOOOLOLOLOLOLOLOLOLOO")
```

```python
say_hello()
```
